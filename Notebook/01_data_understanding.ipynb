# **8. Data Visualization**

Distribusi pendapatan total (AMT_INCOME_TOTAL)

```python
plt.figure(figsize=(10, 6))
sns.histplot(df_1['AMT_INCOME_TOTAL'], bins=30, kde=True)
plt.title('Distribusi Total Pendapatan')
plt.xlabel('Total Pendapatan')
plt.ylabel('Frekuensi')
plt.show()
```

Boxplot untuk pendapatan total

```python
plt.figure(figsize=(10, 6))
sns.boxplot(x=df_1['AMT_INCOME_TOTAL'])
plt.title('Boxplot Total Pendapatan')
plt.xlabel('Total Pendapatan')
plt.show()
```

Distribusi status kredit (STATUS)

```python
plt.figure(figsize=(10, 6))
sns.countplot(x='STATUS', data=df_2)
plt.title('Distribusi Status Kredit')
plt.xlabel('Status Kredit')
plt.ylabel('Jumlah')
plt.show()
```

# **9. Outlier Check**

Menghitung IQR untuk AMT_INCOME_TOTAL

```python
Q1 = df_1['AMT_INCOME_TOTAL'].quantile(0.25)
Q3 = df_1['AMT_INCOME_TOTAL'].quantile(0.75)
IQR = Q3 - Q1
```

Menampilkan outlier

```python
outliers = df_1[(df_1['AMT_INCOME_TOTAL'] < (Q1 - 1.5 * IQR)) | (df_1['AMT_INCOME_TOTAL'] > (Q3 + 1.5 * IQR))]
print("\nOutlier dalam AMT_INCOME_TOTAL:")
print(outliers[['ID', 'AMT_INCOME_TOTAL']])
```
